| **List**                                  | **Set**                                         |

What is short-circuit evaluation?

Short-circuit evaluation is a feature of Python where it stops evaluating a logical expression as soon as the final result is determined. 
With and, evaluation stops when the first condition is False.
With or, evaluation stops when the first condition is True. 
This improves performance and can prevent unnecessary errors, such as division by zero.

What is a ternary operator?

A ternary operator is a one-line conditional expression used as a shortcut for an if...else statement. 
It returns one value if the condition is True and another value if the condition is False.

How are conditions used in analytics?

Conditions (if, elif, else) are used in analytics to make decisions based on data. 
They help classify records, validate data, apply business rules, filter information, detect outliers, and automate decision-making during data analysis.

What happens if no condition is true?

If no condition is True, Python executes the else block if it exists. 
If there is no else block, Python skips the entire conditional statement and continues executing the remaining code.

Can if statement exist without else?

Yes. An if statement can be used without an else. 
If the condition is True, the if block executes. 
If it is False, Python skips the block and continues with the next statement.

What is boolean expression?

A Boolean expression is an expression that evaluates to either True or False.

What is the output of if 0: ?

The condition if 0: evaluates to False because 0 is a falsy value in Python. 
Therefore, the if block is skipped, and only the code after it (or the else block, if present) is executed.

How do you handle multiple conditions?

Multiple conditions in Python can be handled using if, elif, else, logical operators (and, or, not), and nested if statements.

```python
''' Mini Project 1: Sales Transformation Pipeline
Input:
sales = [500, 1200, 3000, 800, 4500]
Tasks:
•	Filter sales > 1000 
•	Apply 10% discount 
•	Calculate total discounted sales 
 '''
from functools import reduce

sales = [500, 1200, 3000, 800, 4500]

filtered_sales = list(filter(lambda x : x>1000,sales))
print("filtered sales :",filtered_sales)
if len(filtered_sales) == 0:
    print("No sales above 1000")
else:

    discount = list(map(lambda x : x - (x*0.10),filtered_sales))
    print("display after discount :",discount)

    total_discounted_sales = reduce(lambda x,y : x+y,discount)
    print("total discounted sales :",total_discounted_sales)
```

```python
# Create generator for numbers
def numbers(n):
    for i in range(1,n+1):
        yield i

gen = numbers(7)

print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
```

```python
#Generate even numbers

def numbers(n):
    for i in range(1,n+1):
        if i%2==0:
            yield i

gen = numbers(14)

print(next(gen))
print(next(gen))
print(next(gen))
```

```python
#Generate squares

def squares(n):
    for i in range(1,n+1):
        yield i**2

gen = squares(7)

print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
```

```python
#Replace hyphens with spaces

import re

text = "manthan-patel-123"
result = re.sub(r"-"," ",text)
print(result)
```

```python
#Validate email address

import re

emails = "abc@gmail.com,manthan.patel@gmail.com,xyz1409@yahoo.com,abc@gmail,@gmail.com,abc@,abc.com,abc@@gmail.com"

pattern = r"[\w.-]+@[\w.-]+\.\w+"

for email in emails.split(","):
    if re.fullmatch(pattern, email):
        print("Valid email id:", email)
    else:
        print("Invalid email id:", email)
```

```python
#Validate date format

import re

dates = [
    "01-01-2026",
    "15-08-2026",
    "25-12-2025",
    "31-03-2024",
    "09-09-2004",
    "1-01-2026",
    "01-1-2026",
    "01-01-26",
    "01/01/2026",
    "2026-01-01"
]

pattern = r"^\d{2}-\d{2}-\d{4}$"

for date in dates:
    if re.fullmatch(pattern, date):
        print("Valid format:", date)
    else:
        print("Invalid format:", date)
```

```python
'''Mini Project 1: Customer Management System
Class Customer:
•	name 
•	age 
•	spend 
Methods:
•	display info 
•	check premium status 
'''

class Customer:
    def __init__(self,name,age,spend):
        self.name = name
        self.age = age
        self.spend = spend

    def info(self):
        return f"customer name: {self.name} , customer age: {self.age} , customer's spend: {self.spend}"

    def premium(self):
        return self.spend>5000

c1 = Customer("manthan",21,2300)
c2 = Customer("aditi",22,5100)
c3 = Customer("meet",23,4100)

print(c1.info())
print("it is primum or not(true/flase):",c1.premium())
print(c2.info())
print("it is primum or not(true/flase):",c2.premium())
print(c3.info())
print("it is primum or not(true/flase):",c3.premium())
```

```python
'''Mini Project 2: Sales Order System
Class Order:
•	product 
•	price 
•	quantity 
Methods:
•	calculate revenue 
•	apply discount 
'''

class Order:
    def __init__(self,product,price,qty):
        self.product = product
        self.price = price
        self.qty = qty

    def ravanue(self):
        return self.price*self.qty
    
    def apply_disscount(self):
        ravanue = self.ravanue()

        if ravanue >= 50000:
            discount = ravanue * 0.10
            return ravanue - discount
        else:
            return ravanue

o1 = Order("phone",12000,3)
o2 = Order("laptop",20000,3)

print("ravanue:",o1.ravanue())
print("after disscount:",o1.apply_disscount())
print("ravanue:",o2.ravanue())
print("after disscount:",o2.apply_disscount())
```

```python
# Create class with private variable
# Access private variable using getter
# Modify private variable using setter

class Employee:                  # Create class with private variable
    def __init__(self,salary):
        self.__salary = salary

    def get_salary(self):        # Access private variable using getter
        return self.__salary

    def set_salary(self,amount): # Modify private variable using setter
        if amount > 0:
            self.__salary = amount

e1 = Employee(2000)
print(e1.get_salary())

e2 = Employee(2000)
e2.set_salary(3000)
print(e2.get_salary())
```

```python
# Create abstract class
# Implement abstract method

from abc import ABC , abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        return "car is starting"

class Bike(Vehicle):
    def start(self):
        return "bike is starting"

car1 = Car()
car1.start()

bike1 = Bike()
bike1.start()
```

```python
# Create customer credit system

class Customer:
    def __init__(self,name,credit_limit):
        self.name = name
        self.__credit_limit = credit_limit
        self.__credit_use = 0

    def use_credit(self,amount):
        if amount < 0:
            print("plz enter valid amount")
        elif self.__credit_use + amount <= self.__credit_limit:
            self.__credit_use += amount 
            print("credit used:",amount)
        else:
            print("limit exceed")

    def make_payment(self,amount):
        if amount < 0:
            print("plz enter valid value")
        elif amount <= self.__credit_use:
            self.__credit_use -= amount
            print("payment sucessfull")
        else:
            print("payment amount is gatter than used amount")

    def get_ava_credit(self):
        return self.__credit_limit - self.__credit_use 

c1 = Customer("manthan",10000)

c1.use_credit(1000)
print("remaing credit:",c1.get_ava_credit())

c1.use_credit(6700)
print("remaing credit:",c1.get_ava_credit())

c1.make_payment(5000)
print("remaing credit:",c1.get_ava_credit())

```

```python
#Build abstract data loader

from abc import ABC , abstractmethod

class DataLoader(ABC):

    @abstractmethod
    def load_data(self):
        pass

class Csvloader(DataLoader):
    def load_data(self):
        return "load data in csv file"

class Jsonloader(DataLoader):
    def load_data(self):
        return "load data in json file"

csv1 = Csvloader()
print(csv1.load_data())

json1 = Jsonloader()
print(json1.load_data())
```

```python
#Explain abstraction via code

from abc import ABC , abstractmethod
#abstract class
class DataLoader(ABC):
    #abstact method
    @abstractmethod
    def load_data(self):
        # Tells WHAT operation is required,
        # but hides HOW it is implemented.
        pass

#child class = provide the implementation
class Csvloader(DataLoader):
    def load_data(self):
        return "load data in csv file"
    
#child class = provide the implementation
class Jsonloader(DataLoader):
    def load_data(self):
        return "load data in json file"

csv1 = Csvloader()
print(csv1.load_data())

json1 = Jsonloader()
print(json1.load_data())

# Abstraction = hide "how it works" and expose "what it does".
```
