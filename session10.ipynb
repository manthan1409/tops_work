```python
# 1.Install Matplotlib in your Python environment and create a simple line plot showing the number of daily steps you took over the last 7 days. Save the chart as steps_lineplot.png.

import matplotlib.pyplot as plt

# Sample daily steps
days = ["Monday", "Tuesday", "Wednesday", "Thursday",
        "Friday", "Saturday", "Sunday"]

steps = [4500, 6000, 7500, 5000, 8500, 10000, 7000]

# Create line plot
plt.plot(days, steps, marker="o", linestyle="-", label="Daily Steps")

# Add title and labels
plt.title("Daily Steps for 7 Days")
plt.xlabel("Days")
plt.ylabel("Number of Steps")

# Add legend and grid
plt.legend()
plt.grid(True)

# Save chart
plt.savefig("steps_lineplot.png", dpi=300, bbox_inches="tight")

# Display chart
plt.show()


# 2.Using Matplotlib, create a scatter plot of 10 restaurants from your city with their Zomato ratings on the x-axis and average meal price on the y-axis. Add axis labels and a title to the chart.

import matplotlib.pyplot as plt

restaurants = [
    "Dominos", "McDonalds", "KFC", "Subway", "Pizza Hut",
    "Honest", "La Pinoz", "Hocco", "Gwalia", "Mocha"
]

ratings = [4.2, 4.0, 4.1, 3.9, 4.3, 4.0, 4.4, 4.1, 3.8, 4.5]

meal_price = [400, 350, 450, 300, 550, 250, 500, 350, 200, 700]

# Create scatter plot
plt.scatter(ratings, meal_price, marker="o", color="blue")

# Add restaurant names beside points
for i in range(len(restaurants)):
    plt.annotate(
        restaurants[i],
        (ratings[i], meal_price[i]),
        fontsize=8
    )

plt.title("Restaurant Ratings vs Average Meal Price")
plt.xlabel("Zomato Rating")
plt.ylabel("Average Meal Price (₹)")

plt.grid(True)
plt.show()

# 3.Build a bar chart that shows the number of orders you or your friends placed on Swiggy, Zomato, and Domino’s in the last month. Use different colors for each bar and add a legend.

import matplotlib.pyplot as plt

apps = ["Swiggy", "Zomato", "Dominos"]

orders = [15, 20, 10]

# Create bar chart
plt.bar(
    apps,
    orders,
    color=["orange", "red", "blue"],
    label="Total Orders"
)

plt.title("Food Delivery Orders in Last Month")
plt.xlabel("Food Delivery Apps")
plt.ylabel("Number of Orders")

plt.legend()
plt.grid(axis="y")

plt.show()

# 4.Create a histogram of the durations (in minutes) of your last 20 Spotify listening sessions using Matplotlib. Set the number of bins to 5 and customize the color of the bars.

import matplotlib.pyplot as plt

# Listening duration in minutes
durations = [
    15, 20, 25, 30, 35,
    40, 45, 50, 55, 60,
    20, 25, 30, 35, 40,
    45, 50, 60, 65, 70
]

# Create histogram
plt.hist(
    durations,
    bins=5,
    color="purple",
    edgecolor="black"
)

plt.title("Spotify Listening Session Duration")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Frequency")

plt.grid(axis="y", alpha=0.5)

plt.show()

# 5.Customize a Matplotlib figure by creating a plot with two subplots: one line plot showing your daily Instagram screen time for a week, and one bar chart showing the number of posts you liked each day. Add appropriate titles, axis labels, and save the figure as social_media_usage.png.<br><br><em><strong>Hint:</strong> Use plt.subplots() to create multiple axes in one figure.</em>

import matplotlib.pyplot as plt

days = ["Monday", "Tuesday", "Wednesday", "Thursday",
        "Friday", "Saturday", "Sunday"]

# Instagram screen time in hours
screen_time = [1.5, 2, 1, 2.5, 3, 4, 3.5]

# Number of posts liked
likes = [20, 30, 15, 35, 40, 50, 45]

# Create two subplots
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# First subplot: Line plot
ax[0].plot(
    days,
    screen_time,
    marker="o",
    color="blue"
)

ax[0].set_title("Daily Instagram Screen Time")
ax[0].set_xlabel("Days")
ax[0].set_ylabel("Screen Time (Hours)")
ax[0].tick_params(axis="x", rotation=45)
ax[0].grid(True)

# Second subplot: Bar chart
ax[1].bar(
    days,
    likes,
    color="orange"
)

ax[1].set_title("Daily Instagram Posts Liked")
ax[1].set_xlabel("Days")
ax[1].set_ylabel("Number of Likes")
ax[1].tick_params(axis="x", rotation=45)
ax[1].grid(axis="y")

# Adjust layout
plt.tight_layout()

# Save figure
plt.savefig("social_media_usage.png", dpi=300, bbox_inches="tight")

# Display figure
plt.show()
```
