```python
# 1.Use plt.subplots() to create a 2x2 grid of subplots and plot four different types of charts (line, bar, scatter, and pie) using any sample data of your choice.

import matplotlib.pyplot as plt
import numpy as np

fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# Line chart
axs[0, 0].plot([1, 2, 3, 4], [10, 20, 25, 30], marker='o', color='blue')
axs[0, 0].set_title('Line Chart')

# Bar chart
axs[0, 1].bar(['A', 'B', 'C', 'D'], [15, 30, 45, 20], color='orange')
axs[0, 1].set_title('Bar Chart')

# Scatter chart
x = np.random.rand(10)
y = np.random.rand(10)
axs[1, 0].scatter(x, y, color='green')
axs[1, 0].set_title('Scatter Chart')

# Pie chart
axs[1, 1].pie([30, 20, 50], labels=['X', 'Y', 'Z'], autopct='%1.1f%%')
axs[1, 1].set_title('Pie Chart')

plt.tight_layout()
plt.show()

# 2.Plot a comparison of average delivery times for Zomato, Swiggy, and Domino's using a bar chart, and style each bar with a different color, linestyle, and linewidth using Matplotlib plot styling options.

import matplotlib.pyplot as plt

companies = ["Zomato", "Swiggy", "Dominos"]
delivery_time = [30, 35, 25]

colors = ["red", "orange", "blue"]
linestyles = ["solid", "dashed", "dotted"]

plt.figure(figsize=(8, 5))

bars = plt.bar(
    companies,
    delivery_time,
    color=colors,
    edgecolor="black",
    linewidth=2
)

# Apply different border styles
for bar, style in zip(bars, linestyles):
    bar.set_linestyle(style)

plt.title("Average Delivery Time Comparison")
plt.xlabel("Food Delivery Companies")
plt.ylabel("Delivery Time (Minutes)")

plt.show()

# 3.Create a multi-axis chart that shows the number of Instagram followers (left y-axis) and average daily posts (right y-axis) for five influencers. Use different colors and markers for each axis.

import matplotlib.pyplot as plt

influencers = ["A", "B", "C", "D", "E"]

followers = [120000, 250000, 180000, 400000, 350000]
daily_posts = [2, 4, 3, 5, 4]

fig, ax1 = plt.subplots(figsize=(9, 5))

# Left Y-axis: Followers
ax1.plot(
    influencers,
    followers,
    color="blue",
    marker="o",
    linewidth=2,
    label="Followers"
)

ax1.set_xlabel("Influencers")
ax1.set_ylabel("Instagram Followers", color="blue")
ax1.tick_params(axis="y", labelcolor="blue")

# Create second Y-axis
ax2 = ax1.twinx()

# Right Y-axis: Daily posts
ax2.plot(
    influencers,
    daily_posts,
    color="red",
    marker="s",
    linewidth=2,
    linestyle="--",
    label="Daily Posts"
)

ax2.set_ylabel("Average Daily Posts", color="red")
ax2.tick_params(axis="y", labelcolor="red")

plt.title("Instagram Followers vs Daily Posts")

plt.show()

# 4.Plot the number of tickets sold for five recent Bollywood movies (categorical) versus their IMDB ratings (numeric) using a scatter plot. Add annotations to display the movie names above each point.

import matplotlib.pyplot as plt

movies = ["Jawan", "Pathaan", "Animal", "Dangal", "Stree 2"]

tickets_sold = [120, 150, 110, 180, 140]
imdb_ratings = [7.0, 5.8, 6.1, 8.3, 7.0]

plt.figure(figsize=(9, 6))

plt.scatter(
    tickets_sold,
    imdb_ratings,
    color="purple",
    s=120,
    marker="o"
)

# Add movie names above each point
for i in range(len(movies)):
    plt.annotate(
        movies[i],
        (tickets_sold[i], imdb_ratings[i]),
        xytext=(0, 10),
        textcoords="offset points",
        ha="center"
    )

plt.title("Bollywood Movies: Tickets Sold vs IMDb Ratings")
plt.xlabel("Tickets Sold (Illustrative Units)")
plt.ylabel("IMDb Ratings")
plt.grid(True)

plt.show()

# 5.Add a custom text annotation to a Matplotlib chart showing Flipkart's monthly sales, marking the highest sales point with the label 'Big Billion Days'.<br><br><em><strong>Hint:</strong> Use the ax.annotate() function to place the label at the correct data point.</em>

import matplotlib.pyplot as plt

months = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

sales = [120, 150, 130, 180, 200, 170,
         220, 250, 280, 350, 600, 300]

# Find highest sales
max_sales = max(sales)
max_index = sales.index(max_sales)

plt.figure(figsize=(11, 6))

plt.plot(
    months,
    sales,
    marker="o",
    color="blue",
    linewidth=2
)

# Highlight highest sales
plt.annotate(
    "Big Billion Days",
    xy=(months[max_index], max_sales),
    xytext=(months[max_index], max_sales + 100),
    arrowprops=dict(
        arrowstyle="->",
        color="red",
        linewidth=2
    ),
    fontsize=12,
    color="red"
)

plt.title("Flipkart Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales (Illustrative Units)")
plt.grid(True)

plt.show()
```
