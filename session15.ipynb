```python
# 1.Install the pandas-profiling library and generate a profile report for the 'Spotify Top 100 Songs' dataset (CSV available on Kaggle); open the HTML report and note any missing values or data type issues.
# 2.Use Sweetviz to create a comparison report between two CSV files: one containing Zomato restaurant data for Mumbai and another for Delhi. Briefly describe one key difference you spot in the visual report.<br><br><em><strong>Hint:</strong> Use analyze() for single file and compare() for two datasets.</em>

import pandas as pd

mumbai = pd.read_csv(r"C:\Users\manthan\Downloads\zomato_mumbai.csv")
delhi = pd.read_csv(r"C:\Users\manthan\Downloads\zomato_delhi.csv")

print("Mumbai Dataset:")
print(mumbai.head())

print("\nDelhi Dataset:")
print(delhi.head())

import sweetviz as sv

report = sv.compare(
    [mumbai, "Mumbai"],
    [delhi, "Delhi"]
)

report.show_html("zomato_comparison.html", open_browser=True)

print("Comparison report generated successfully!")

print("Mumbai Average Rating:")
print(mumbai["rating"].mean())

print("\nDelhi Average Rating:")
print(delhi["rating"].mean())

# 3.Open the Myntra product listings dataset in D-Tale, explore the interface, and use it to filter products with a price above ₹2000. Take a screenshot of the filtered view and note the number of such products.

import pandas as pd

df = pd.read_csv(r"C:\Users\manthan\Downloads\myntra_product_details.csv")

print(df.head())
print(df.shape)
print(df.columns)
print(df["price"].head())
df["price"] = (
    df["price"]
    .astype("string")
    .str.replace(r"[₹,]", "", regex=True)
)

df["price"] = pd.to_numeric(df["price"], errors="coerce")
filtered_df = df[df["price"] > 2000]

print(filtered_df.head(10))

print("Products above ₹2000:", len(filtered_df))
import dtale

d = dtale.show(df)
d.open_browser()

# 4.Given a pandas-profiling report for a Flipkart product reviews dataset, interpret and list two columns that may need cleaning or transformation before further analysis.<br><br><em><strong>Hint:</strong> Look for columns with high cardinality, missing values, or warnings in the report.</em>

import pandas as pd

df = pd.read_csv(r"C:\Users\manthan\Downloads\flipkart_product_reviews.csv")

print(df.head())
print("Missing Values:")
print(df.isnull().sum().sort_values(ascending=False))
print("Unique Values:")
print(df.nunique().sort_values(ascending=False))
print("Data Types:")
print(df.dtypes)
from ydata_profiling import ProfileReport

profile = ProfileReport(
    df,
    title="Flipkart Reviews Analysis",
    explorative=True
)

profile.to_file("flipkart_report.html")

# 5.Pick any one auto-EDA tool (pandas-profiling, Sweetviz, or D-Tale) and use ChatGPT or Copilot to generate a code snippet that loads a Swiggy food order CSV and produces a summary report. Run the code and attach the generated report or a screenshot as proof.

df = pd.read_csv(r"C:\Users\manthan\Downloads\swiggy_orders.csv")
import pandas as pd

df = pd.read_csv("swiggy_orders.csv")

print("First five rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)
from ydata_profiling import ProfileReport

profile = ProfileReport(
    df,
    title="Swiggy Food Orders EDA Report",
    explorative=True
)

profile.to_file("swiggy_eda_report.html")

print("Swiggy EDA Report Generated Successfully!")
import os
import webbrowser

path = os.path.abspath("swiggy_eda_report.html")

webbrowser.open("file://" + path)
```
