```python
## 2.Find a small JSON dataset of trending songs (for example, from Spotify's API samples or any open JSON file), and use pd.read_json() to import it into a DataFrame. Display the DataFrame's column names and data types using df.info(). 
import json

songs = [
    {
        "song": "Blinding Lights",
        "artist": "The Weeknd",
        "popularity": 95,
        "streams": 1000000
    },
    {
        "song": "Shape of You",
        "artist": "Ed Sheeran",
        "popularity": 90,
        "streams": 900000
    },
    {
        "song": "Perfect",
        "artist": "Ed Sheeran",
        "popularity": 85,
        "streams": 750000
    },
    {
        "song": "Levitating",
        "artist": "Dua Lipa",
        "popularity": 88,
        "streams": 800000
    },
    {
        "song": "Starboy",
        "artist": "The Weeknd",
        "popularity": 87,
        "streams": 820000
    }
]

with open("trending_songs.json", "w") as file:
    json.dump(songs, file, indent=4)

print("JSON file created successfully!")
```

```python
# 4.Download a large Excel file (at least 10,000 rows) of Flipkart product listings. Use pd.read_excel() with the chunksize parameter to read the file in chunks of 2000 rows, and print the number of rows in each chunk as you iterate.



import pandas as pd

# Load the CSV file
df = pd.read_csv(
    r"C:\Users\manthan\Downloads\flipkart_products_dataset_sample.csv"
)

chunk_size = 2000

# Process the DataFrame in chunks of 2000 rows
for start in range(0, len(df), chunk_size):
    chunk = df.iloc[start:start + chunk_size]
    print("Rows in this chunk:", len(chunk))

print("Total rows:", len(df))

```
