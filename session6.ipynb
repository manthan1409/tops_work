```python
# 1.Download a sample CSV file of IPL player stats (include columns like player name, runs, matches, and some missing values). Load it into a pandas DataFrame and use isnull() and notnull() to print how many missing values are present in each column.

import pandas as pd

# Create sample IPL player data
data = {
    "player_name": [
        "Virat Kohli", "Rohit Sharma", "MS Dhoni",
        "Jasprit Bumrah", "KL Rahul", "Hardik Pandya"
    ],
    "matches": [15, 14, None, 12, 14, 10],
    "runs": [741, None, 161, 25, 520, None]
}

df = pd.DataFrame(data)

# Save as CSV
df.to_csv("ipl_player_stats.csv", index=False)

# Load the CSV file
df = pd.read_csv("ipl_player_stats.csv")

print(df)

# Count missing values in each column
print("Missing values:")
print(df.isnull().sum())

# Count non-missing values in each column
print("\nNon-missing values:")
print(df.notnull().sum())

# 2.Using the loaded IPL player stats DataFrame, apply dropna(axis=0, how='any') to remove all rows with any missing data and display the shape of the DataFrame before and after dropping.

# Shape before dropping missing values
print("Before dropping:", df.shape)

# Remove rows containing at least one missing value
df_clean = df.dropna(axis=0, how="any")

# Shape after dropping missing values
print("After dropping:", df_clean.shape)

print(df_clean)

# 3.For the 'runs' column in your IPL player stats DataFrame, use fillna() to replace missing values with the mean of the column. Print the updated column to verify the changes.

# Calculate the mean of the runs column
mean_runs = df["runs"].mean()

print("Mean runs:", mean_runs)

# Fill missing runs with the mean
df["runs"] = df["runs"].fillna(mean_runs)

print("\nUpdated runs column:")
print(df["runs"])

# 4.Simulate a Zomato-style restaurant ratings dataset with some missing ratings. Use forward fill (method='ffill') to fill missing values in the ratings column, then use backward fill (method='bfill') for any remaining missing values. Show the before and after results.

import pandas as pd

# Create sample restaurant ratings
ratings_df = pd.DataFrame({
    "restaurant": ["Restaurant A", "Restaurant B", "Restaurant C",
                   "Restaurant D", "Restaurant E", "Restaurant F"],
    "rating": [4.2, None, 3.8, None, None, 4.5]
})

print("Before filling:")
print(ratings_df)

# Forward fill
ratings_df["rating"] = ratings_df["rating"].ffill()

# Backward fill any remaining missing values
ratings_df["rating"] = ratings_df["rating"].bfill()

print("\nAfter filling:")
print(ratings_df)

# 5.Pick any one column with missing values from your dataset and explain which missingness mechanism (MCAR, MAR, or MNAR) is most likely and why, using 2-3 sentences.<br><br><em><strong>Hint:</strong> Think about whether the missing data is random or related to other factors in the dataset.</em>

# For the rating column in the restaurant dataset, MAR (Missing At Random) may be the most likely mechanism if rating availability is related to an observed factor, such as restaurant location or price category. For example, restaurants in a particular location might have fewer recorded ratings. However, we cannot determine the true missingness mechanism from the missing values alone; we need additional information about why ratings were not recorded.
```
